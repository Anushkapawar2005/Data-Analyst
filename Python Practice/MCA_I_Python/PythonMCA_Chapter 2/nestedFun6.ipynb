```python
def outer_fun(fname):
    print('I am in outer function...')
    print(f'Your first name is {fname}')
    def inner_fun(lname):
        print("And now..., I am in inner function...")
        print(f'Your last name is {lname}')
    ln = input("Enter your last name: ")
    inner_fun(ln)

fn = input("Enter your first name: ")
outer_fun(fn)
```

```python
def talk(line):
    def say(word):
        print(word)
    words= line.split(' ')
    for word in words:
        say(word)

talk("THis is a line")
```
