```python
def mygenerator():
    print('First Item')
    yield 10
    print('Second item')
    yield 20
    print('Third item')
    yield 30
    print('Fourth item')
    yield 40
    print('Five item')
    yield 50
gen = mygenerator()
```

```python
next(gen)
```

```python
for item in gen:
    print(item)
```

```python
while True:
    try:
        item=next(gen)
        print(item)
    except StopIteration:
        break
```

```python
def table_gen(num):
    for i in range(1,11):
        yield f'{num} X {i} = {num*i}'
n = int(input("Enter any no for table generation: "))
table = table_gen(n)
type(table)
for item in table:
    print(item)
```

```python
def OddEven_gen(nums):
    for item in nums:
        if item%2 == 0:
            yield f'{item} is even'
        else:
            yield f'{item} is odd'
nl=[int(i) for i in input("Enter numbers separated by comma: ").split(',')]
obj = OddEven_gen(nl)
for item in obj:
    print(item)
```

```python
def square_seq(num):
    for i in range(1,num+1):
        yield f'square of {i} = {i**2}'

n= int(input("Enter a number to get squares of sequence upto:"))
seq = square_seq(n)
while True:
    try:
        item= next(seq)
        print(item)
    except StopIteration:
        break
```

```python
def evens_of_squar(x):
    for item in range(1,x+1):
        if item%2==0:
            yield f'Square of Even number {item} = {item**2}'
n=int(input("Enter a number to get squares of even numbers upto :"))
even = evens_of_squar(n)
for i in even:
    print(i)
```
