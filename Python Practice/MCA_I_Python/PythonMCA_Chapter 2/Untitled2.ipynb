```python
def nums(*num):
    sum=0
    for i in num:
        sum+=i
    print("Sum:",sum)
def mul(*num):
    mult=1
    for i in num:
        mult*=i
    print("multiplication:",mult)
nums(1,2,3,4,5)
mul(2,3,4,5)
```
