```python
# map() : applies the specified fun to every item of the passed iterable, yeild results and return an iterator
# syntax : map(function, iterable) -> 
```

```python
help(map)
```

```python
l = [int(i) for i in input("Enter numbers separated by comma: ").split(',')]
print(l)
square = list(map(lambda x: x*x, l))
print(square)
```

```python
square = list(map(lambda x: x*x, [int(i) for i in range(1,11)]))
print(square)
for i in range(len(square)):
    print(f'Square of {i+1}={square[i]}')
```

```python
l = [int(i) for i in input("Enter numbers separated by comma: ").split(',')]
print(l)
cube=list(map(lambda x:x*x*x,l))
for i in range(len(cube)):
    print(f'Cube of {l[i]}={cube[i]}')
```

```python
num = int(input("Enter any num for table..:"))
table = list(map(lambda x: x*num,[int(i) for i in range(1,11)]))
print(table)
for i in range(len(table)):
    print(f'{num} x {i+1} = {table[i]}')
```

```python
num = [int(input("Enter any num for table..:"))]
table = list(map(lambda x,y : x*y, [int(i) for i in range(1,11)], num*10))

print(num)
print(table)

for i in range(len(table)):
    print(f'{num[0]} x {i+1} = {table[i]}')
```
