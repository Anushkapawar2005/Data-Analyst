```python
def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def div(a,b):
    return a/b

def pow(a,b):
    return a**b


def main():
    while True:
        print("\n Menu: ")
        print("1. Addition")
        print("2. Substraction")
        print("3. Multiplication")
        print("4. Division")
        print("5. Power")

        choice = int(input("Enter your choice: "))
        
        a = int(input("Enter a: "))
        b = int(input("ENter b: "))

        if choice == 1:
            print("Addition: ", add(a,b))
        elif choice == 2:
            print("Substraction: ", sub(a,b))
        elif choice == 4:
            print("Multiplication: ", mul(a,b))
        elif choice == 5:
            print("Power: ", pow(a,b))
        else:
            print("Invalid choice!")
            ans = input("Do you want to continue ? (yes/no): ")

            if ans.lower() != "yes":
                print("Program terminated.")
                break

if __name__ == "__main__":
    main()
```
