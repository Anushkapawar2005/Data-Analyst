```python
def greet(name,fn):
    print(f'hello {name}')
    fn()
fn = lambda: print("Good morning")
greet('Anu',fn)
```

```python
odd = lambda x: x%2!=0
even = lambda x: x%2==0
print(odd(11))
print(even(13))
```

```python
def oddEven(num, odd):
    if odd(num) == True:
        print(f'{num} is odd number..')
    else:
        print(f'{num} is even number..')
odd = lambda x: x%2 != 0
oddEven(23,odd)
```

```python
def EvenOdd(num, even):
    if even(num) == True:
        print(f'{num} is even number')
    else:
        print(f'{num} is odd number')
even = lambda x: x%2==0
EvenOdd(22,even)
```
