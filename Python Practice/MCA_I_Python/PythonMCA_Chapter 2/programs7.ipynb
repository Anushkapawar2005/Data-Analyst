```python
#prime number

def isprime(num):
    if num<=1:
        return False
    for i in range(2,num):
        if num%i==0:
            return False
    return True

num = int(input("Enter a number: "))

if isprime(num):
    print(num, "is a not a prime no")
            
else:
    print(num,"it is a prime number")
            
```

```python
# Factorial
num = int(input("Enter num:"))
fact=1
for i in range(1,num+1):
    fact*=i
print("Factorial of ",num,"=",fact)
```

```python
# Factorial
num = int(input("Enter num:"))
fact=1
for i in range(num,0,-1):
    print(f'{i} * {fact} = {i*fact}')
    fact*=i
print("Factorial of ",num," = ",fact)
```

```python
num = int(input("Enter num:"))
fact=1
i=num
while i>0:
    print(f'{i} * {fact} = {i*fact}')
    fact*=i
    i-=1
print("Factorial of ",num," = ",fact)
```

```python
# Factorial
def fact(num):
    
    fact=1
    for i in range(num,0,-1):
        print(f'{i} * {fact} = {i*fact}')
        fact*=i
    return fact

num = int(input("Enter num:"))
f = fact(num)
print(f'Fcatorial of {num} is: {f}')
    
```

```python
# Recursion function
def rec_fact(n):
    if n <= 1:
        return 1
    else:
        return (n* rec_fact(n-1))

        
num = int(input("Enter num:"))
if num < 0:
    print("Factorial Doesn't exist for negative number: ")
elif num==0:
    print("Factorial of 0 is 1...")
else:
    print(f'Factorial of {num} is {rec_fact(num)}')
```
