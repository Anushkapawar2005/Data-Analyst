```python
def person(f_name,l_name):
    print(f'Hello {l_name} {f_name}')
    print(f'Hello {f_name} {l_name}')

person(f_name='anushka',l_name='pawar')
person('Seeta','Shinde')
person(l_name='pawar',f_name='anushka')
```

```python
# keyword argument **kwarg
def personDetails(**person):
    print(f'Hello {person['f_name']} ,{person['l_name']} and age is {person['age']}')

personDetails(f_name='anu', l_name='pawar',age=21)
personDetails(l_name='pawar', f_name='anu',age=21)
personDetails(f_name='anu', l_name='pawar',age=21, sal=4000)
```

```python
#parameter with default value
def d_value(name,msg='Good morning'):
    print(f'Hello {name}....{msg}')
d_value('anu','Good evening')
d_value('seeta')
```
