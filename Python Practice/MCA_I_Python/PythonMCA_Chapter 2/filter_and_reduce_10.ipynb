```python
help(filter)
```

```python
l = [int(i) for i in input("Enter numbers separated by comma: ").split(',')]
evenlist = list(filter(lambda x: x%2 == 0, l))
print("Even list = ",evenlist)
oddlist = list(filter(lambda x: x%2 != 0, l))
print("Odd list: ",oddlist)
```

```python
years = [int(i) for i in input("Enter years: ").split(',')]
leapyears = list(filter(lambda x: ((x%4 == 0 and x%100 != 0) or (x%400 == 0)),years))
print("leap years: ",leapyears)
```

```python
# Python reduce function
```

```python
from functools import reduce
```

```python
help(reduce)
```

```python
sum = reduce(lambda x,y:x+y, [1,2,3,4,5])
print(sum)
```

```python
mult = reduce(lambda x,y:x*y,[1,2,3,4,5])
print(mult)
```

```python
max = reduce(lambda x,y: x if x>y else y , [12,3,2,15,85,23])
print(max)
```

```python
min = reduce(lambda x,y: x if x<y else y , [12,3,2,15,85,23])
print(min)
```

```python
# find sum of squares 
sum_squares = reduce(lambda x,y: x + (y*y),[2,3,4],0 )
print(sum_squares)
```
