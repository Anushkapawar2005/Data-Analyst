```python
# 13. Write a Python program to sort (ascending and descending) a dictionary by value.
```

```python
d = {'a': 3, 'b': 1, 'c': 4, 'd': 2}

items = list(d.items())

for i in range(len(items)):
    for j in range(i+1, len(items)):
        if items[i][1] > items[j][1]:
            items[i], items[j] = items[j], items[i]

print("Original Dictionary:", d)
print("Ascending order: ", dict(items))
print("Descending order: ", dict(items[::-1]))
```

```python
# 38. Write a Python program to check the validity of password input by users.
#    Validation:
#    - At least 1 letter between [a-z] and 1 letter between [A-Z].
#    - At least 1 number between [0-9].
#    - At least 1 character from [$#@].
#    - Minimum length 6 characters.
#    - Maximum length 16 characters.
```

```python
password = input("Enter Password: ")
lower=False
upper=False
digit=False
special=False

for ch in password:
    if ch >= 'a' and ch<='z':
        lower = True
    elif ch >= 'A' and ch<='Z':
        upper = True
    elif ch >= '0' and ch <= '9':
        digit = True
    elif ch == '$' or ch == '#' or ch == '@':
        special = True

if (len(password)>=6 and len(password)<=16 and lower and upper and digit and special):
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
# 45. A Lower triangular matrix is a square matrix (where the number of rows and columns are equal) where all the elements 
# above the diagonal are zero.
```

```python
# 46. Write a Python program to convert a square matrix into an upper triangular matrix.
```

```python

n = int(input("Enter size of matrix: "))

matrix = []

for i in range(n):
    row = []
    for j in range(n):
        num = int(input("Enter element: "))
        row.append(num)
    matrix.append(row)

for i in range(n):
    for j in range(n):
        if i > j:
            matrix[i][j] = 0

print("Upper Triangular Matrix:")
for i in range(n):
    for j in range(n):
        print(matrix[i][j], end=" ")
    print()
```

```python
# 47. Write a Python program to check whether a given matrix is a Binary Matrix or not.
```

```python

n, m = input("Enter rows and columns: ").split()
n = int(n)
m = int(m)

matrix = []

for i in range(n):
    row = []
    elements = input("Enter row elements: ").split()

    for j in range(m):
        row.append(int(elements[j]))

    matrix.append(row)

binary = True

for i in range(n):
    for j in range(m):
        if matrix[i][j] != 0 and matrix[i][j] != 1:
            binary = False

if binary:
    print("YES")
else:
    print("NO")
```

```python
# 49.Given a list of n-1 numbers ranging from 1 to n, your task is to find the missing number. There are no duplicates
```

```python

numbers = input("Enter numbers: ").split()

for i in range(len(numbers)):
    numbers[i] = int(numbers[i])

n = len(numbers) + 1
total = n * (n + 1) // 2
missing = total - sum(numbers)

print("Missing number:", missing)
```
